# Run the operator tests
The same tests CI runs on every pull request: set `OPERATOR_FILE` below to the operator you are working on, then run the last cell. The tests load that operator file (from the project root or `operators/`) and run its logic on rows where we already know the answer.

```python
# Change this to the YAML file of the operator you want to test.
# tests/test_operators.py looks for it at the project root and in operators/.
OPERATOR_FILE = "business-day-counter-v1.0.0.yaml"
```

```python
import os, sys
import pytest

sys.dont_write_bytecode = True
os.environ["OPERATOR_FILE"] = OPERATOR_FILE  # picked up by tests/test_operators.py
print(f"Testing operator: {OPERATOR_FILE}")

# Drop the cached test module so edits to tests/test_operators.py are
# picked up when this cell runs again in the same session.
for name in [m for m in sys.modules if "test_operators" in m]:
    del sys.modules[name]

exit_code = pytest.main(["-v", "-p", "no:cacheprovider", os.path.join(os.getcwd(), "tests")])
assert exit_code == 0, "Some operator tests failed, see the lines marked FAILED above"
print("All operator tests passed")
```
